Una **función** es un bloque de código con un nombre que realiza una tarea concreta. Python trae muchas funciones ya incluidas (*built-in functions*) que se pueden usar directamente. En este notebook se explican las funciones de esta categoría: qué hacen, su sintaxis, sus parámetros y ejemplos prácticos con su resultado.

**Fuente principal:** [Documentación oficial de Python – Built-in Functions](https://docs.python.org/3/library/functions.html)

### 10. `list()`
**Descripción:** Crea una lista (colección ordenada y **modificable**). También convierte otros iterables en lista.

**Sintaxis:** `list(iterable)`

**Parámetros:** `iterable`: texto, tupla, rango, conjunto, etc.

```python
print(list("hola"))              # ['h', 'o', 'l', 'a']
print(list((1, 2, 3)))           # [1, 2, 3]
print(list(range(5)))            # [0, 1, 2, 3, 4]

frutas = list()                  # lista vacía
frutas.append("manzana")
frutas.append("pera")
print(frutas)
```

### 11. `tuple()`
**Descripción:** Crea una tupla (colección ordenada **no modificable**).

**Sintaxis:** `tuple(iterable)`

**Parámetros:** `iterable`: texto, lista, rango, etc.

```python
t = tuple([1, 2, 3])
print(t)                         # (1, 2, 3)
print(tuple("abc"))              # ('a', 'b', 'c')
print(t[0])                      # se puede leer
# t[0] = 99  -> daría TypeError (no se puede modificar)
```

### 12. `dict()`
**Descripción:** Crea un diccionario: una colección de pares **clave → valor**.

**Sintaxis:** `dict(**kwargs)  o  dict(iterable_de_pares)`

**Parámetros:** Pares clave=valor, o una lista de tuplas (clave, valor).

```python
persona = dict(nombre="Luis", edad=30)
print(persona)
print(persona["nombre"])

pares = dict([("a", 1), ("b", 2)])
print(pares)

persona["ciudad"] = "Bogotá"     # agregar una clave nueva
print(persona)
```

### 13. `set()`
**Descripción:** Crea un conjunto: colección **sin elementos repetidos** y sin orden.

**Sintaxis:** `set(iterable)`

**Parámetros:** `iterable`: lista, texto, tupla, etc.

```python
numeros = [1, 2, 2, 3, 3, 3]
print(set(numeros))              # {1, 2, 3}  (elimina repetidos)
print(set("banana"))             # letras únicas

a = {1, 2, 3}
b = {3, 4, 5}
print(a & b)                     # intersección: {3}
print(a | b)                     # unión: {1, 2, 3, 4, 5}
```
