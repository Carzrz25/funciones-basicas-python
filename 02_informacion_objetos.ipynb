Una **función** es un bloque de código con un nombre que realiza una tarea concreta. Python trae muchas funciones ya incluidas (*built-in functions*) que se pueden usar directamente. En este notebook se explican las funciones de esta categoría: qué hacen, su sintaxis, sus parámetros y ejemplos prácticos con su resultado.

### 3. `len()`
**Descripción:** Devuelve la cantidad de elementos de un objeto (texto, lista, tupla, diccionario, conjunto, etc.).

**Sintaxis:** `len(objeto)`

**Parámetros:** `objeto`: la secuencia o colección a medir.

```python
print(len("Python"))                       # 6 caracteres
print(len([10, 20, 30, 40]))              # 4 elementos
print(len({"a": 1, "b": 2}))              # 2 pares clave-valor
print(len(""))                            # 0
```

### 4. `type()`
**Descripción:** Devuelve el tipo de dato de un objeto.

**Sintaxis:** `type(objeto)`

**Parámetros:** `objeto`: el valor del que se quiere conocer el tipo.

```python
print(type(5))
print(type(3.14))
print(type("hola"))
print(type([1, 2]))
print(type(True))
print(type(None))
```

### 5. `isinstance()`
**Descripción:** Comprueba si un objeto es de un tipo determinado. Devuelve `True` o `False`.

**Sintaxis:** `isinstance(objeto, tipo)`

**Parámetros:** `objeto`: el valor a revisar; `tipo`: el tipo (o una tupla de tipos) con el que se compara.

```python
print(isinstance(5, int))                 # True
print(isinstance("5", int))               # False
print(isinstance(3.5, (int, float)))      # True: es int O float

valor = [1, 2, 3]
if isinstance(valor, list):
    print("Es una lista con", len(valor), "elementos")
```
