### 30. `open()`
**Descripción:** Abre un archivo para leerlo o escribirlo. Lo recomendable es usarla con `with`, que cierra el archivo automáticamente.

**Sintaxis:** `open(archivo, modo='r', encoding=None)`

**Parámetros:** `archivo`: nombre o ruta; `modo`: `'r'` leer, `'w'` escribir (borra lo anterior), `'a'` agregar al final; `encoding`: codificación (usa `'utf-8'` para tildes).

```python
# Escribir un archivo
with open("ejemplo.txt", "w", encoding="utf-8") as f:
    f.write("Primera línea\n")
    f.write("Segunda línea con tildes: canción\n")

# Agregar contenido al final
with open("ejemplo.txt", "a", encoding="utf-8") as f:
    f.write("Tercera línea\n")

# Leer todo el archivo
with open("ejemplo.txt", "r", encoding="utf-8") as f:
    contenido = f.read()
print(contenido)

# Leer línea por línea
with open("ejemplo.txt", "r", encoding="utf-8") as f:
    for linea in f:
        print("->", linea.strip())
```
