Una **función** es un bloque de código con un nombre que realiza una tarea concreta. Python trae muchas funciones ya incluidas (*built-in functions*) que se pueden usar directamente. En este notebook se explican las funciones de esta categoría: qué hacen, su sintaxis, sus parámetros y ejemplos prácticos con su resultado.

**Fuente principal:** [Documentación oficial de Python – Built-in Functions](https://docs.python.org/3/library/functions.html)

### 6. `int()`
**Descripción:** Convierte un valor a número entero. Si recibe un decimal, lo trunca (quita la parte decimal sin redondear).

**Sintaxis:** `int(valor, base=10)`

**Parámetros:** `valor`: número o texto a convertir; `base`: base numérica si el valor es un texto.

```python
print(int("42"))          # 42
print(int(7.9))           # 7  (trunca, no redondea)
print(int("1010", 2))     # 10 (binario a decimal)
# int("hola")  -> daría ValueError
```

### 7. `float()`
**Descripción:** Convierte un valor a número decimal (punto flotante).

**Sintaxis:** `float(valor)`

**Parámetros:** `valor`: número o texto a convertir.

```python
print(float("3.14"))      # 3.14
print(float(5))           # 5.0
print(float("2e3"))       # 2000.0 (notación científica)
```

### 8. `str()`
**Descripción:** Convierte cualquier valor a texto.

**Sintaxis:** `str(objeto)`

**Parámetros:** `objeto`: el valor que se quiere convertir a texto.

```python
edad = 20
mensaje = "Tengo " + str(edad) + " años"
print(mensaje)
print(str(3.5), str(True), str([1, 2]))
print(type(str(100)))
```

### 9. `bool()`
**Descripción:** Convierte un valor a booleano (`True` o `False`). Son **falsos**: `0`, `0.0`, `""`, `[]`, `{}`, `()`, `set()` y `None`. Todo lo demás es verdadero.

**Sintaxis:** `bool(valor)`

**Parámetros:** `valor`: el elemento a evaluar.

```python
print(bool(1))        # True
print(bool(0))        # False
print(bool("hola"))   # True
print(bool(""))       # False
print(bool([]))       # False
print(bool([0]))      # True (la lista tiene un elemento)
```
