Una **función** es un bloque de código con un nombre que realiza una tarea concreta. Python trae muchas funciones ya incluidas (*built-in functions*) que se pueden usar directamente. En este notebook se explican las funciones de esta categoría: qué hacen, su sintaxis, sus parámetros y ejemplos prácticos con su resultado.

**Fuente principal:** [Documentación oficial de Python – Built-in Functions](https://docs.python.org/3/library/functions.html)

### 1. `print()`
**Descripción:** Muestra en pantalla uno o varios valores.

**Sintaxis:** `print(*objetos, sep=' ', end='\n')`

**Parámetros:** `objetos`: lo que se quiere mostrar; `sep`: separador entre los valores; `end`: lo que se imprime al final (por defecto un salto de línea).

```python
print("Hola, mundo")
print("Python", "es", "genial")
print("a", "b", "c", sep="-")
print("Sin salto de línea", end=" ")
print("continúa aquí")
nombre = "Ana"
edad = 20
print(f"{nombre} tiene {edad} años")
```

### 2. `input()`
**Descripción:** Pide al usuario que escriba un dato por teclado. **Siempre devuelve un texto (`str`)**.

**Sintaxis:** `input(mensaje)`

**Parámetros:** `mensaje`: texto opcional que se muestra al usuario antes de pedir el dato.

```python
# Uso real (en un script normal se ejecutaría así):
# nombre = input("¿Cómo te llamas? ")
# print("Hola,", nombre)

# Para que el notebook no se detenga, simulamos lo que el usuario escribiría:
nombre = "Camila"   # equivale a lo que escribiría el usuario
edad_texto = "25"   # input() siempre entrega texto
print("Hola,", nombre)
print(type(edad_texto))                 # <class 'str'>
print("El próximo año tendrás", int(edad_texto) + 1)
```
