Una **función** es un bloque de código con un nombre que realiza una tarea concreta. Python trae muchas funciones ya incluidas (*built-in functions*) que se pueden usar directamente. En este notebook se explican las funciones de esta categoría: qué hacen, su sintaxis, sus parámetros y ejemplos prácticos con su resultado.

**Fuente principal:** [Documentación oficial de Python – Built-in Functions](https://docs.python.org/3/library/functions.html)

### 26. `map()`
**Descripción:** Aplica una función a **cada elemento** de un iterable y devuelve los resultados.

**Sintaxis:** `map(funcion, iterable)`

**Parámetros:** `funcion`: la función a aplicar; `iterable`: los elementos sobre los que se aplica.

```python
numeros = [1, 2, 3, 4]
cuadrados = list(map(lambda x: x ** 2, numeros))
print(cuadrados)                           # [1, 4, 9, 16]

textos = ["10", "20", "30"]
print(list(map(int, textos)))              # [10, 20, 30]
print(list(map(str.upper, ["a", "b"])))    # ['A', 'B']
```

### 27. `filter()`
**Descripción:** Se queda solo con los elementos que **cumplen una condición** (la función devuelve `True`).

**Sintaxis:** `filter(funcion, iterable)`

**Parámetros:** `funcion`: condición a evaluar; `iterable`: elementos a filtrar.

```python
numeros = [1, 2, 3, 4, 5, 6]
pares = list(filter(lambda x: x % 2 == 0, numeros))
print(pares)                               # [2, 4, 6]

palabras = ["sol", "luna", "estrella", "mar"]
largas = list(filter(lambda p: len(p) > 3, palabras))
print(largas)                              # ['luna', 'estrella']
```

### 28. `any()`
**Descripción:** Devuelve `True` si **al menos un** elemento del iterable es verdadero.

**Sintaxis:** `any(iterable)`

**Parámetros:** `iterable`: elementos a evaluar.

```python
print(any([False, False, True]))     # True
print(any([0, "", None]))            # False
print(any([]))                       # False (lista vacía)

edades = [15, 17, 22]
print(any(e >= 18 for e in edades))  # True: hay al menos un mayor de edad
```

### 29. `all()`
**Descripción:** Devuelve `True` si **todos** los elementos del iterable son verdaderos.

**Sintaxis:** `all(iterable)`

**Parámetros:** `iterable`: elementos a evaluar.

```python
print(all([True, True, True]))       # True
print(all([True, False, True]))      # False
print(all([]))                       # True (lista vacía)

notas = [4.0, 3.5, 4.8]
print(all(n >= 3.0 for n in notas))  # True: todos aprobaron
```
