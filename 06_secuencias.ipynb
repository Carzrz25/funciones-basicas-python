### 21. `range()`
**Descripción:** Genera una secuencia de números enteros. Se usa mucho en ciclos `for`.

**Sintaxis:** `range(fin)  |  range(inicio, fin)  |  range(inicio, fin, paso)`

**Parámetros:** `inicio`: primer valor (por defecto 0); `fin`: valor donde se detiene (**no se incluye**); `paso`: incremento (por defecto 1).

```python
print(list(range(5)))           # [0, 1, 2, 3, 4]
print(list(range(2, 8)))        # [2, 3, 4, 5, 6, 7]
print(list(range(0, 20, 5)))    # [0, 5, 10, 15]
print(list(range(10, 0, -3)))   # [10, 7, 4, 1]

for i in range(1, 4):
    print("Vuelta número", i)
```

### 22. `sorted()`
**Descripción:** Devuelve una **nueva lista ordenada** a partir de un iterable (no modifica el original).

**Sintaxis:** `sorted(iterable, key=None, reverse=False)`

**Parámetros:** `iterable`: elementos a ordenar; `key`: función que define el criterio; `reverse`: `True` para orden descendente.

```python
numeros = [5, 2, 9, 1]
print(sorted(numeros))                  # [1, 2, 5, 9]
print(sorted(numeros, reverse=True))    # [9, 5, 2, 1]
print(numeros)                          # el original no cambia

palabras = ["uva", "manzana", "kiwi"]
print(sorted(palabras))                 # alfabético
print(sorted(palabras, key=len))        # por longitud
```

### 23. `reversed()`
**Descripción:** Devuelve un iterador que recorre la secuencia **al revés**.

**Sintaxis:** `reversed(secuencia)`

**Parámetros:** `secuencia`: lista, tupla, texto o rango.

```python
print(list(reversed([1, 2, 3, 4])))    # [4, 3, 2, 1]
print("".join(reversed("Python")))      # 'nohtyP'

for n in reversed(range(1, 4)):
    print("Cuenta regresiva:", n)
```

### 24. `enumerate()`
**Descripción:** Recorre un iterable devolviendo **el índice y el valor** al mismo tiempo.

**Sintaxis:** `enumerate(iterable, start=0)`

**Parámetros:** `iterable`: lo que se recorre; `start`: número desde el que empieza el contador.

```python
colores = ["rojo", "verde", "azul"]
for indice, color in enumerate(colores):
    print(indice, color)

print("---")
for numero, color in enumerate(colores, start=1):
    print(f"{numero}. {color}")
```

### 25. `zip()`
**Descripción:** Une varios iterables elemento a elemento, creando tuplas. Se detiene cuando se acaba el más corto.

**Sintaxis:** `zip(iterable1, iterable2, ...)`

**Parámetros:** Dos o más iterables.

```python
nombres = ["Ana", "Luis", "Eva"]
notas = [4.5, 3.9, 5.0]

for nombre, nota in zip(nombres, notas):
    print(nombre, "->", nota)

print(list(zip(nombres, notas)))
print(dict(zip(nombres, notas)))   # crear diccionario fácilmente
```
