Una **función** es un bloque de código con un nombre que realiza una tarea concreta. Python trae muchas funciones ya incluidas (*built-in functions*) que se pueden usar directamente. En este notebook se explican las funciones de esta categoría: qué hacen, su sintaxis, sus parámetros y ejemplos prácticos con su resultado.

**Fuente principal:** [Documentación oficial de Python – Built-in Functions](https://docs.python.org/3/library/functions.html)

### 14. `abs()`
**Descripción:** Devuelve el valor absoluto de un número (su distancia a cero, siempre positivo).

**Sintaxis:** `abs(numero)`

**Parámetros:** `numero`: entero o decimal.

```python
print(abs(-8))        # 8
print(abs(5))         # 5
print(abs(-3.14))     # 3.14
```

### 15. `round()`
**Descripción:** Redondea un número a una cantidad de decimales.

**Sintaxis:** `round(numero, ndigitos=None)`

**Parámetros:** `numero`: el valor a redondear; `ndigitos`: cantidad de decimales (si se omite, devuelve un entero).

```python
print(round(3.14159, 2))   # 3.14
print(round(7.6))          # 8
print(round(1234, -2))     # 1200 (redondea a centenas)
print(round(2.5))          # 2  (Python redondea .5 al par más cercano)
```

### 16. `pow()`
**Descripción:** Calcula una potencia: base elevada a un exponente. Equivale al operador `**`.

**Sintaxis:** `pow(base, exponente, modulo=None)`

**Parámetros:** `base`: número base; `exponente`: potencia; `modulo`: opcional, devuelve el resto de la división.

```python
print(pow(2, 3))        # 8
print(2 ** 3)           # 8, es lo mismo
print(pow(9, 0.5))      # 3.0 (raíz cuadrada)
print(pow(2, 10, 1000)) # 24 -> (2**10) % 1000
```

### 17. `divmod()`
**Descripción:** Devuelve una tupla con el **cociente** y el **residuo** de una división entera.

**Sintaxis:** `divmod(a, b)`

**Parámetros:** `a`: dividendo; `b`: divisor.

```python
print(divmod(17, 5))          # (3, 2) porque 17 = 5*3 + 2
cociente, residuo = divmod(17, 5)
print("Cociente:", cociente, "| Residuo:", residuo)

# Ejemplo práctico: convertir 135 minutos a horas y minutos
horas, minutos = divmod(135, 60)
print(f"{horas} h {minutos} min")
```

### 18. `sum()`
**Descripción:** Suma todos los elementos de un iterable de números.

**Sintaxis:** `sum(iterable, inicio=0)`

**Parámetros:** `iterable`: lista, tupla, etc. con números; `inicio`: valor que se suma al total (opcional).

```python
notas = [4.5, 3.8, 5.0, 4.2]
print(sum(notas))                 # 17.5
print(sum(notas) / len(notas))    # promedio
print(sum([1, 2, 3], 10))         # 16 (empieza en 10)
```

### 19. `min()`
**Descripción:** Devuelve el elemento más pequeño.

**Sintaxis:** `min(iterable)  o  min(a, b, c, ...)`

**Parámetros:** Un iterable o varios valores sueltos; `key` opcional para elegir el criterio de comparación.

```python
print(min(4, 9, 2))                       # 2
print(min([10, 3, 8]))                    # 3
print(min("python"))                      # 'h' (orden alfabético)
print(min(["kiwi", "uva", "manzana"], key=len))   # 'uva' (la más corta)
```

### 20. `max()`
**Descripción:** Devuelve el elemento más grande.

**Sintaxis:** `max(iterable)  o  max(a, b, c, ...)`

**Parámetros:** Un iterable o varios valores sueltos; `key` opcional para elegir el criterio de comparación.

```python
print(max(4, 9, 2))                       # 9
print(max([10, 3, 8]))                    # 10
print(max(["kiwi", "uva", "manzana"], key=len))   # 'manzana' (la más larga)

edades = {"Ana": 20, "Luis": 25, "Eva": 22}
print(max(edades, key=edades.get))        # 'Luis' (quien tiene mayor edad)
```
